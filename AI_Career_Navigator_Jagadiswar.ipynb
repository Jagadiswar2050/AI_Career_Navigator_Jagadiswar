### Goal
Build a focused career-advice assistant using layered prompts: **role + instructions + few-shot examples**. The project then demonstrates how the prompt structure can be connected to a lightweight local response engine.

## 1. Prompt Design Strategy

### Role Prompt
Defines the assistant as a practical career mentor.

### Instruction Prompt
Controls the response format: track, skills, roadmap, project and next step.

### Few-Shot Prompt
Provides examples so the expected structure is clear before a new question is processed.

```python
ROLE_PROMPT = '''\nYou are CareerNavigator, a professional and practical career mentor.\nGive concise, structured, beginner-friendly guidance.\nAvoid invented facts.\nPrefer: Goal -> Skills -> Roadmap -> Project -> Next Step.\n'''\n\nINSTRUCTION_PROMPT = '''\nResponse rules:\n- Understand career intent\n- Identify a relevant track\n- Give a short roadmap\n- Recommend practical skills and one project\n- Use bullet points\n'''
```

```python
FEW_SHOT_EXAMPLES = '''\nUser: I want to become a data analyst.\nAssistant: Learn Python + SQL, practice visualization, build a dashboard.\n\nUser: I want to enter cybersecurity.\nAssistant: Learn networking + Linux, study security basics, use authorized labs.\n'''\n\ndef build_prompt(user_question):\n    return ROLE_PROMPT + '\\n' + INSTRUCTION_PROMPT + '\\n' + FEW_SHOT_EXAMPLES + \\\n           f'\\nUser: {user_question}\\nAssistant:'\n\nprint(build_prompt('I want to learn AI'))
```

## 2. Local Career Response Engine
A small rule-based engine is used so the notebook runs without external APIs or keys. The prompt itself is still documented and assembled to demonstrate prompt-engineering principles.

```python
def career_response(user_text):\n    user = user_text.lower()\n    if any(k in user for k in ['data analyst', 'analytics']):\n        return 'Track: Data Analytics\\n- Learn Python + SQL\\n- Practice visualization\\n- Build a dashboard'\n    if any(k in user for k in ['ai', 'machine learning', 'ml']):\n        return 'Track: AI / Machine Learning\\n- Learn Python + NumPy + Pandas\\n- Study ML basics\\n- Build a prediction project'\n    if any(k in user for k in ['cyber', 'security']):\n        return 'Track: Cybersecurity\\n- Learn networking + Linux\\n- Study security fundamentals\\n- Practice in authorized labs'\n    if any(k in user for k in ['resume', 'cv']):\n        return 'Track: Resume & Portfolio\\n- Keep it concise\\n- Add 1–2 projects\\n- Highlight evidence-based skills'\n    return 'Track: General Career Planning\\n- Identify a target role\\n- Learn core skills\\n- Build a project\\n- Improve resume and communication'\n\ndef career_navigator(question):\n    return career_response(question)
```

```python
queries = [\n    'I want to learn AI',\n    'How do I become a data analyst?',\n    'How can I start cybersecurity?',\n    'How should I improve my resume?'\n]\n\nfor q in queries:\n    print('USER:', q)\n    print(career_navigator(q))\n    print('-' * 50)
```
